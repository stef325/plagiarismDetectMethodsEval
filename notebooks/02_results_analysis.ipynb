# Inspeção dos Resultados do Experimento

Este notebook ajuda a abrir rapidamente os principais artefatos já gerados pelo experimento: CSVs, relatórios e algumas estatísticas consolidadas.

```python
from pathlib import Path
import pandas as pd
from IPython.display import Markdown, display

def find_existing_project() -> Path | None:
    candidates = [Path.cwd(), *Path.cwd().parents]
    for candidate in candidates:
        if (candidate / "src" / "main.py").exists():
            return candidate
    return None

PROJECT_DIR = find_existing_project()
if PROJECT_DIR is None:
    raise FileNotFoundError("Abra este notebook dentro do repositorio do projeto.")

RESULTS_DIR = PROJECT_DIR / "data" / "results"
RESULTS_DIR
```

```python
paths = {
    "similarity": RESULTS_DIR / "experiment" / "similarity_results.csv",
    "robustness": RESULTS_DIR / "evaluation" / "robustness_metrics.csv",
    "interpretability": RESULTS_DIR / "evaluation" / "interpretability" / "interpretability_results.csv",
    "summary": RESULTS_DIR / "consolidated" / "experiment_summary.csv",
    "statistics": RESULTS_DIR / "consolidated" / "statistics_summary.csv",
    "report": RESULTS_DIR / "consolidated" / "consolidated_results.md",
}

for name, path in paths.items():
    print(f"{name}: {path.exists()} -> {path}")
```

```python
if paths["similarity"].exists():
    similarity_df = pd.read_csv(paths["similarity"])
    display(similarity_df.head())
    print("Quantidade de comparacoes:", len(similarity_df))
    if "pair_type" in similarity_df.columns:
        display(similarity_df["pair_type"].value_counts())
else:
    print("Arquivo similarity_results.csv nao encontrado.")
```

```python
if paths["robustness"].exists():
    robustness_df = pd.read_csv(paths["robustness"])
    display(robustness_df)
else:
    print("Arquivo robustness_metrics.csv nao encontrado.")
```

```python
if paths["interpretability"].exists():
    interpretability_df = pd.read_csv(paths["interpretability"])
    display(interpretability_df.head())
    if "transformation" in interpretability_df.columns:
        display(interpretability_df["transformation"].value_counts().head(20))
else:
    print("Arquivo interpretability_results.csv nao encontrado.")
```

```python
if paths["summary"].exists():
    summary_df = pd.read_csv(paths["summary"])
    display(summary_df)
else:
    print("Arquivo experiment_summary.csv nao encontrado.")

if paths["statistics"].exists():
    statistics_df = pd.read_csv(paths["statistics"])
    display(statistics_df.head(20))
else:
    print("Arquivo statistics_summary.csv nao encontrado.")
```

```python
if paths["report"].exists():
    display(Markdown(paths["report"].read_text(encoding="utf-8")))
else:
    print("Relatorio consolidado nao encontrado.")
```

```python
figures_dir = RESULTS_DIR / "figures"
if figures_dir.exists():
    figure_files = sorted(figures_dir.rglob("*.png"))
    print(f"Quantidade de figuras PNG: {len(figure_files)}")
    for figure in figure_files[:20]:
        print(figure.relative_to(PROJECT_DIR))
else:
    print("Diretorio de figuras nao encontrado.")
```
